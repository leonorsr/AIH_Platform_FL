# EDA — Icentia11k em C#

20 clientes: **16 treino, 2 dev e 2 teste**. Contagens por classe, qualidade, heterogeneidade entre clientes e exemplos de sinais preparados.

Antes de abrir, compilar na raiz: `dotnet build src/Aih.Ecg -c Release`. Executar com o kernel **.NET (C#)** / Polyglot Notebooks, a partir desta pasta. O código da análise está em `src/Aih.Ecg/Exploration.cs`.


```
#r "../src/Aih.Ecg/bin/Release/net8.0/Aih.Ecg.dll"
using System;
using System.IO;
using System.Text.Json;
using Aih.Ecg;

var root = Dataset.FindRoot();
var data = Path.Combine(root, "data");
Console.WriteLine(data);
```

```
var reportDir = Path.Combine(root, "artifacts", "eda");
var result = Exploration.Run(data, reportDir);
Console.WriteLine(JsonSerializer.Serialize(result, new JsonSerializerOptions { WriteIndented = true }));
```

## Consultar resultados

Abrir o ficheiro `artifacts/eda/eda.html` no navegador para ver as tabelas e os gráficos. Os CSV permitem reutilizar as contagens. A análise lê os índices de forma sequencial e carrega apenas um sinal de treino por classe.

Q representa batimentos não classificáveis. As janelas de 1 segundo têm 500 pontos interpolados, normalizados por batimento. Dev/test permanecem reservados para avaliação; não usar o teste para ajustar parâmetros de treino.


```
Console.WriteLine(File.ReadAllText(Path.Combine(reportDir, "split_class_counts.csv")));
Console.WriteLine(File.ReadAllText(Path.Combine(reportDir, "quality_counts.csv")));
```
