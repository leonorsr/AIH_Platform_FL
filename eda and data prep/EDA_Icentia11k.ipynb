# EDA — Icentia11k in C#

20 clients: **16 training, 2 development, and 2 test clients**. Explore class counts, quality, client heterogeneity, and prepared-signal examples.

Before opening, build from the repository root: `dotnet build src/Aih.Ecg -c Release`. Use the **.NET (C#)** / Polyglot Notebooks kernel from this directory. Analysis code is implemented in `src/Aih.Ecg/Exploration.cs`.


```
#r "../src/Aih.Ecg/bin/Release/net8.0/Aih.Ecg.dll"
using System;
using System.IO;
using System.Text.Json;
using Aih.Ecg;

var root = Dataset.FindRoot();
var data = Path.Combine(root, "data");
Console.WriteLine(data);
```

```
var reportDir = Path.Combine(root, "artifacts", "eda");
var result = Exploration.Run(data, reportDir);
Console.WriteLine(JsonSerializer.Serialize(result, new JsonSerializerOptions { WriteIndented = true }));
```

## View results

Open `artifacts/eda/eda.html` in a browser to view the tables and charts. CSV files provide reusable counts. Analysis reads indexes sequentially and loads only one training signal per class.

Q means unclassifiable. One-second windows contain 500 interpolated points, normalized per beat. Development/test patients are reserved for evaluation; do not use test results to tune training parameters.


```
Console.WriteLine(File.ReadAllText(Path.Combine(reportDir, "split_class_counts.csv")));
Console.WriteLine(File.ReadAllText(Path.Combine(reportDir, "quality_counts.csv")));
```
