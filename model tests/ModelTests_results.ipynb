# Smoke-test results

## Data partition

The 20 patients retain their fixed partition: **16 for training, 2 for development and 2 for testing**, with no patients shared between splits. The held-out test set was not used.

These smoke tests used **5 547 training beats** and **1 294 development beats**, with the same sample for every network. Sampling was stratified by class and patient: up to 128 examples per class/patient for training and 256 for development, without replacement. Each network was trained with **3 seeds and up to 3 epochs**. For each seed, the checkpoint with the highest development S/V PR-AUC is used. Stratification changes prevalence; these results are for model screening.
